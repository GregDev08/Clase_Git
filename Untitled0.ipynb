````python
import numpy as np
import builtins

notas_recuperacion = [3.0, 9.5, 4.2, 8.8, 5.0, 9.1, 3.5, 7.7, 4.8, 6.0]
calificaciones_array = np.array(notas_recuperacion)

builtins.print(f"Array de calificaciones: {calificaciones_array}")
media_calificaciones = np.mean(calificaciones_array)
builtins.print(f"Media de las calificaciones: {media_calificaciones:.2f}")

# Respuesta a la pregunta 2:
# ¿Es la media siempre un buen resumen del grupo? ¿Por qué sí o por qué no?
# La media no siempre es el mejor resumen, especialmente si hay valores atípicos (outliers).
# Por ejemplo, en el caso de los salarios con un outlier de 25000, la media se ve
# muy afectada y no representa bien la mayoría de los salarios. Para las calificaciones,
# si no hay outliers extremos, la media puede ser un buen indicador del rendimiento general.
````

````python
## 🌡️ Reto — Desviación estándar · Sensores de una fábrica
Trabajas en el equipo técnico de una fábrica de chocolate 🍫.
La temperatura del horno debe mantenerse estable en torno a 175°C.
Tenéis dos sensores midiendo la temperatura cada hora durante un día:
````

````python
import numpy as np
import builtins
sensor_A = np.array([174, 176, 175, 174, 176, 175, 175, 174, 176, 175,
                     174, 176, 175, 175, 174, 176, 175, 174, 175, 176,
                     175, 174, 176, 175])
sensor_B = np.array([168, 182, 171, 179, 165, 185, 172, 178, 169, 181,
                     174, 176, 167, 183, 170, 180, 166, 184, 173, 177,
                     168, 182, 175, 175])
builtins.print(f"Sensor A → Media: {np.mean(sensor_A):.1f}°C")
builtins.print(f"Sensor B → Media: {np.mean(sensor_B):.1f}°C")
````

````python
import builtins
import numpy as np
from scipy import stats

sensor_A = np.array([174, 176, 175, 174, 176, 175, 175, 174, 176, 175,
                     174, 176, 175, 175, 174, 176, 175, 174, 175, 176,
                     175, 174, 176, 175])
sensor_B = np.array([168, 182, 171, 179, 165, 185, 172, 178, 169, 181,
                     174, 176, 167, 183, 170, 180, 166, 184, 173, 177,
                     168, 182, 175, 175])
median_A = np.median(sensor_A)
median_B = np.median(sensor_B)

mode_A = stats.mode(sensor_A)
mode_B = stats.mode(sensor_B)

media_A = np.mean(sensor_A)
media_B = np.mean(sensor_B)

desviacion_A = np.std(sensor_A)
desviacion_B = np.std(sensor_B)

builtins.print(f"Sensor A → Media: {media_A:.1f}°C")
builtins.print(f"Sensor B → Media: {media_B:.1f}°C")
builtins.print(f"Sensor A → Mediana: {median_A:.1f}°C")
builtins.print(f"Sensor B → Mediana: {median_B:.1f}°C")
builtins.print(f"Sensor A → Moda: {mode_A.mode}°C")
builtins.print(f"Sensor B → Moda: {mode_B.mode}°C")
builtins.print(f"Sensor A → std: {desviacion_A:.2f}")
builtins.print(f"Sensor B → std: {desviacion_B:.2f}")
````
